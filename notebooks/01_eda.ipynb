## 4. Visual EDA - Numerical Feature Analysis

```python
# Select numerical features
numerical_features = df.select_dtypes(include=['int64', 'float64']).columns

print("Numerical Features:")
print(list(numerical_features))
print("\nNumber of Numerical Features:", len(numerical_features))
```

```python
# Statistical summary of numerical features
df[numerical_features].describe().T
```

### 4.1 Distribution of Numerical Features

```python
import matplotlib.pyplot as plt

# Plot histograms for numerical features
df[numerical_features].hist(
    figsize=(16, 14),
    bins=30
)

plt.suptitle("Distribution of Numerical Features", fontsize=16)
plt.tight_layout()
plt.show()
```

### 4.2 Boxplot Analysis for Numerical Features

```python
# Create boxplots for numerical features

plt.figure(figsize=(16, 10))

df[numerical_features].boxplot(rot=90)

plt.title("Boxplots of Numerical Features")
plt.ylabel("Values")
plt.tight_layout()
plt.show()
```

### 4.3 Correlation Heatmap

```python
# Calculate correlation matrix
correlation_matrix = df[numerical_features.tolist() + ['Revenue']].corr()

# Plot correlation heatmap
plt.figure(figsize=(14, 10))

plt.imshow(correlation_matrix, cmap='coolwarm', aspect='auto')
plt.colorbar(label='Correlation')

plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    rotation=90
)

plt.yticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

### 4.4 Categorical Features vs Revenue

```python
categorical_features = df.select_dtypes(
    include=['object', 'bool']
).columns

print("Categorical / Boolean Features:")
print(list(categorical_features))
```

```python
# Purchase rate for each categorical feature

for feature in categorical_features:
    print(f"\n--- {feature} ---")
    
    purchase_rate = (
        df.groupby(feature)['Revenue']
        .mean()
        .sort_values(ascending=False)
        * 100
    )
    
    print(purchase_rate)
```

### 4.5 Outlier Analysis using IQR

```python
# Detect outliers using the IQR method

outlier_summary = []

for feature in numerical_features:
    Q1 = df[feature].quantile(0.25)
    Q3 = df[feature].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[
        (df[feature] < lower_bound) |
        (df[feature] > upper_bound)
    ]
    
    outlier_summary.append({
        'Feature': feature,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outlier Count': len(outliers),
        'Outlier %': (len(outliers) / len(df)) * 100
    })

outlier_df = pd.DataFrame(outlier_summary)

outlier_df.sort_values(
    by='Outlier Count',
    ascending=False
)
```

```python
# Plot outlier counts

outlier_plot = outlier_df.sort_values(
    by='Outlier Count',
    ascending=True
)

plt.figure(figsize=(10, 7))

plt.barh(
    outlier_plot['Feature'],
    outlier_plot['Outlier Count']
)

plt.title("Outlier Count by Numerical Feature")
plt.xlabel("Number of Outliers")
plt.ylabel("Feature")

plt.tight_layout()
plt.show()
```

### 4.6 Important Features Related to Revenue

```python
# Calculate correlation of numerical features with Revenue

revenue_correlation = (
    correlation_matrix['Revenue']
    .drop('Revenue')
    .sort_values(key=abs, ascending=False)
)

print("Features ranked by correlation with Revenue:")
print(revenue_correlation)
```

```python
# Visualize feature correlations with Revenue

plt.figure(figsize=(10, 7))

revenue_correlation.sort_values().plot(kind='barh')

plt.title("Feature Correlation with Revenue")
plt.xlabel("Correlation")
plt.ylabel("Feature")

plt.tight_layout()
plt.show()
```

```python
# Calculate purchase-rate difference for categorical features

categorical_summary = []

for feature in categorical_features:
    
    rates = (
        df.groupby(feature)['Revenue']
        .mean() * 100
    )
    
    categorical_summary.append({
        'Feature': feature,
        'Minimum Purchase Rate (%)': rates.min(),
        'Maximum Purchase Rate (%)': rates.max(),
        'Difference (%)': rates.max() - rates.min()
    })

categorical_importance = pd.DataFrame(categorical_summary)

categorical_importance.sort_values(
    by='Difference (%)',
    ascending=False
)
```


## 4.7 EDA Summary

The exploratory data analysis revealed several important patterns in the Online Shoppers Purchasing Intention dataset:

* The dataset contains **12,330 records and 18 columns**, with **no missing values**.
* There are **125 duplicate records**, which should be handled during the preprocessing stage.
* The target variable `Revenue` is imbalanced, with **15.47% purchase sessions** and **84.53% non-purchase sessions**.
* `PageValues` has the **strongest positive correlation with Revenue (0.493)**, indicating that sessions with higher page-value measurements are more strongly associated with purchases.
* `ExitRates` has the **strongest negative correlation with Revenue (-0.207)**, followed by `BounceRates (-0.151)`.
* `ProductRelated` and `ProductRelated_Duration` have positive correlations with Revenue (**0.159** and **0.152**, respectively), suggesting that product-page activity may be associated with purchase behavior.
* `Administrative` also shows a positive correlation with Revenue (**0.139**).
* `Informational`, `Administrative_Duration`, and `Informational_Duration` show relatively weak positive correlations with Revenue.
* `OperatingSystems`, `Browser`, `Region`, and `TrafficType` show very weak linear correlations with Revenue.
* Histogram and boxplot analysis indicate that several numerical features are **right-skewed** and contain potential outliers.
* The categorical feature analysis shows that purchase rates vary across different categories of `Month`, `VisitorType`, and `Weekend`.

Overall, the EDA suggests that **page-value, exit-rate, bounce-rate, and product-related browsing behavior** are among the more informative features for predicting whether an online shopping session will result in a purchase. These findings will guide the **data preprocessing, feature engineering, and model development** stages.
