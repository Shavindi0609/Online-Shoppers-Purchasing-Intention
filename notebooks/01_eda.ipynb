```python
print(df['Revenue'].value_counts())
```

```python
target_percentage = df['Revenue'].value_counts(normalize=True) * 100

print(target_percentage)
```

```python
plt.figure(figsize=(6, 4))

sns.countplot(data=df, x='Revenue')

plt.title('Revenue Distribution')
plt.xlabel('Revenue')
plt.ylabel('Number of Sessions')

plt.show()
```

```python
revenue_labels = df['Revenue'].map({
    False: 'No Purchase',
    True: 'Purchase'
})

plt.figure(figsize=(6, 4))

sns.countplot(x=revenue_labels)

plt.title('Purchase Intention Distribution')
plt.xlabel('Session Outcome')
plt.ylabel('Number of Sessions')

plt.show()
```
